```python
# Step 1: Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 2: Set random seed for reproducibility
np.random.seed(42)

# Step 3: Define sample size
n = 50

# Step 4: Create simulated dataset
data = pd.DataFrame({
    # Production Output
    "Units_Manual": np.random.randint(45, 60, n),
    "Units_Human_AI": np.random.randint(70, 90, n),

    # Error Rate (%)
    "Error_Manual": np.random.randint(7, 12, n),
    "Error_Human_AI": np.random.randint(2, 5, n),

    # Downtime (minutes)
    "Downtime_Manual": np.random.randint(110, 150, n),
    "Downtime_Human_AI": np.random.randint(50, 80, n),

    # Cost per Unit (₹)
    "Cost_Manual": np.random.randint(115, 140, n),
    "Cost_Human_AI": np.random.randint(85, 110, n)
})

# Step 5: Display first 10 rows
print("Sample Data:\n")
print(data.head(10))

# Step 6: Save dataset
data.to_csv("manufacturing_efficiency_dataset.csv", index=False)

# Step 7: Calculate averages
avg_manual = data[["Units_Manual", "Error_Manual", "Downtime_Manual", "Cost_Manual"]].mean()
avg_ai = data[["Units_Human_AI", "Error_Human_AI", "Downtime_Human_AI", "Cost_Human_AI"]].mean()

print("\nAverage Manual System:\n", avg_manual)
print("\nAverage Human-AI System:\n", avg_ai)

# Step 8: Calculate improvement %
productivity_improvement = ((avg_ai["Units_Human_AI"] - avg_manual["Units_Manual"]) / avg_manual["Units_Manual"]) * 100
error_reduction = ((avg_manual["Error_Manual"] - avg_ai["Error_Human_AI"]) / avg_manual["Error_Manual"]) * 100
downtime_reduction = ((avg_manual["Downtime_Manual"] - avg_ai["Downtime_Human_AI"]) / avg_manual["Downtime_Manual"]) * 100
cost_reduction = ((avg_manual["Cost_Manual"] - avg_ai["Cost_Human_AI"]) / avg_manual["Cost_Manual"]) * 100

print("\n--- Performance Improvement ---")
print("Productivity Improvement (%):", round(productivity_improvement, 2))
print("Error Reduction (%):", round(error_reduction, 2))
print("Downtime Reduction (%):", round(downtime_reduction, 2))
print("Cost Reduction (%):", round(cost_reduction, 2))

# Step 9: Visualization

# Production Comparison
plt.figure()
plt.bar(["Manual", "Human-AI"], [avg_manual["Units_Manual"], avg_ai["Units_Human_AI"]])
plt.title("Production Output Comparison")
plt.ylabel("Units per Hour")
plt.show()

# Error Rate Comparison
plt.figure()
plt.bar(["Manual", "Human-AI"], [avg_manual["Error_Manual"], avg_ai["Error_Human_AI"]])
plt.title("Error Rate Comparison")
plt.ylabel("Error (%)")
plt.show()

# Downtime Comparison
plt.figure()
plt.bar(["Manual", "Human-AI"], [avg_manual["Downtime_Manual"], avg_ai["Downtime_Human_AI"]])
plt.title("Downtime Comparison")
plt.ylabel("Minutes")
plt.show()

# Cost Comparison
plt.figure()
plt.bar(["Manual", "Human-AI"], [avg_manual["Cost_Manual"], avg_ai["Cost_Human_AI"]])
plt.title("Cost per Unit Comparison")
plt.ylabel("Cost (₹)")
plt.show()
```

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Set seed for reproducibility
np.random.seed(42)

# Number of samples
n = 500

# Create simulated dataset
data = pd.DataFrame({
    # Production Output
    "Units_Manual": np.random.randint(45, 60, n),
    "Units_Human_AI": np.random.randint(70, 95, n),

    # Error Rates
    "Error_Manual (%)": np.random.uniform(6, 10, n).round(2),
    "Error_Human_AI (%)": np.random.uniform(2, 5, n).round(2),

    # Downtime
    "Downtime_Manual (min)": np.random.randint(110, 150, n),
    "Downtime_Human_AI (min)": np.random.randint(40, 80, n),

    # Cost
    "Cost_Manual (₹)": np.random.randint(115, 135, n),
    "Cost_Human_AI (₹)": np.random.randint(85, 105, n)
})

# Display first 10 rows
print("Sample Data:\n")
print(data.head(10))

# Save dataset
data.to_csv("manufacturing_efficiency_500_samples.csv", index=False)

# Calculate averages
print("\nAverage Values:\n")
print(data.mean())

# Calculate improvements
prod_improvement = ((data["Units_Human_AI"].mean() - data["Units_Manual"].mean())
                    / data["Units_Manual"].mean()) * 100

error_reduction = ((data["Error_Manual (%)"].mean() - data["Error_Human_AI (%)"].mean())
                   / data["Error_Manual (%)"].mean()) * 100

downtime_reduction = ((data["Downtime_Manual (min)"].mean() - data["Downtime_Human_AI (min)"].mean())
                      / data["Downtime_Manual (min)"].mean()) * 100

cost_reduction = ((data["Cost_Manual (₹)"].mean() - data["Cost_Human_AI (₹)"].mean())
                  / data["Cost_Manual (₹)"].mean()) * 100

print("\nPerformance Improvements:\n")
print(f"Production Improvement: {prod_improvement:.2f}%")
print(f"Error Reduction: {error_reduction:.2f}%")
print(f"Downtime Reduction: {downtime_reduction:.2f}%")
print(f"Cost Reduction: {cost_reduction:.2f}%")

# Plot comparison
labels = ["Production", "Error", "Downtime", "Cost"]
manual_means = [
    data["Units_Manual"].mean(),
    data["Error_Manual (%)"].mean(),
    data["Downtime_Manual (min)"].mean(),
    data["Cost_Manual (₹)"].mean()
]

ai_means = [
    data["Units_Human_AI"].mean(),
    data["Error_Human_AI (%)"].mean(),
    data["Downtime_Human_AI (min)"].mean(),
    data["Cost_Human_AI (₹)"].mean()
]

x = np.arange(len(labels))

plt.figure()
plt.bar(x - 0.2, manual_means, width=0.4, label="Manual")
plt.bar(x + 0.2, ai_means, width=0.4, label="Human-AI")
plt.xticks(x, labels)
plt.title("Operational Efficiency Comparison")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Set seed for reproducibility
np.random.seed(42)

# Number of samples
n_samples = 500

# -----------------------------
# Simulated Workload Model
# -----------------------------
# Workload scale: 0 (low) to 100 (high)

# Traditional ATC workload (higher mean, more variability)
traditional_workload = np.random.normal(loc=75, scale=10, size=n_samples)

# Human-AI collaborative workload (lower mean, less variability)
ai_collab_workload = np.random.normal(loc=55, scale=8, size=n_samples)

# Clip values to realistic range
traditional_workload = np.clip(traditional_workload, 0, 100)
ai_collab_workload = np.clip(ai_collab_workload, 0, 100)

# -----------------------------
# Metrics Calculation
# -----------------------------
mean_traditional = np.mean(traditional_workload)
mean_ai = np.mean(ai_collab_workload)

std_traditional = np.std(traditional_workload)
std_ai = np.std(ai_collab_workload)

# Workload reduction percentage
reduction_percent = ((mean_traditional - mean_ai) / mean_traditional) * 100

# -----------------------------
# Output Results
# -----------------------------
print("===== Workload Comparison =====")
print(f"Samples: {n_samples}")
print(f"Traditional ATC Mean Workload: {mean_traditional:.2f}")
print(f"AI Collaborative Mean Workload: {mean_ai:.2f}")
print(f"Traditional Std Dev: {std_traditional:.2f}")
print(f"AI Collaborative Std Dev: {std_ai:.2f}")
print(f"Workload Reduction: {reduction_percent:.2f}%")

# -----------------------------
# Visualization
# -----------------------------
plt.figure()

plt.hist(traditional_workload, bins=30, alpha=0.6, label='Traditional ATC')
plt.hist(ai_collab_workload, bins=30, alpha=0.6, label='AI Collaborative')

plt.xlabel("Workload Level")
plt.ylabel("Frequency")
plt.title("Workload Distribution Comparison (500 Samples)")
plt.legend()

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Set seed for reproducibility
np.random.seed(42)

# Number of samples
n_samples = 500

# Generate synthetic driving conditions
traffic_density = np.random.uniform(0, 1, n_samples)   # 0 (low) to 1 (high)
weather_condition = np.random.uniform(0, 1, n_samples) # 0 (clear) to 1 (severe)
road_complexity = np.random.uniform(0, 1, n_samples)   # 0 (simple) to 1 (complex)

# Error modeling
# Fully Automated Systems (higher sensitivity to complexity)
auto_error = (
    0.4 * traffic_density +
    0.35 * weather_condition +
    0.45 * road_complexity +
    np.random.normal(0, 0.05, n_samples)
)

# Human-AI Shared Control (reduced error due to human intervention)
shared_error = (
    0.25 * traffic_density +
    0.20 * weather_condition +
    0.30 * road_complexity +
    np.random.normal(0, 0.04, n_samples)
)

# Clip errors to non-negative values
auto_error = np.clip(auto_error, 0, None)
shared_error = np.clip(shared_error, 0, None)

# Create dataset
data = pd.DataFrame({
    "Traffic Density": traffic_density,
    "Weather Condition": weather_condition,
    "Road Complexity": road_complexity,
    "Auto Error": auto_error,
    "Shared Control Error": shared_error
})

# Compute statistics
mean_auto_error = np.mean(auto_error)
mean_shared_error = np.mean(shared_error)
error_reduction = ((mean_auto_error - mean_shared_error) / mean_auto_error) * 100

# Print results
print("===== ERROR ANALYSIS =====")
print(f"Mean Error (Fully Automated): {mean_auto_error:.4f}")
print(f"Mean Error (Shared Control): {mean_shared_error:.4f}")
print(f"Error Reduction (%): {error_reduction:.2f}%")

# Visualization
plt.figure()
plt.hist(auto_error, bins=30, alpha=0.5, label='Fully Automated Error')
plt.hist(shared_error, bins=30, alpha=0.5, label='Shared Control Error')
plt.xlabel("Error Magnitude")
plt.ylabel("Frequency")
plt.title("Error Distribution Comparison (500 Samples)")
plt.legend()
plt.show()

# Scatter comparison
plt.figure()
plt.scatter(auto_error, shared_error, alpha=0.6)
plt.xlabel("Fully Automated Error")
plt.ylabel("Shared Control Error")
plt.title("Error Correlation Between Systems")
plt.show()

# Save dataset
data.to_csv("autonomous_error_comparison_500_samples.csv", index=False)
```

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Set seed for reproducibility
np.random.seed(42)

# Number of samples
n = 500

# Create simulated dataset
data = pd.DataFrame({
    # Diagnostic Accuracy (%)
    "Accuracy_Conventional (%)": np.random.uniform(80, 90, n).round(2),
    "Accuracy_Human_AI (%)": np.random.uniform(90, 98, n).round(2),

    # Error Rate (%)
    "Error_Conventional (%)": np.random.uniform(10, 20, n).round(2),
    "Error_Human_AI (%)": np.random.uniform(2, 8, n).round(2),

    # False Positives
    "FalsePos_Conventional": np.random.randint(5, 15, n),
    "FalsePos_Human_AI": np.random.randint(1, 8, n),

    # False Negatives
    "FalseNeg_Conventional": np.random.randint(5, 12, n),
    "FalseNeg_Human_AI": np.random.randint(1, 6, n),

    # Diagnosis Time (minutes)
    "Time_Conventional (min)": np.random.uniform(8, 15, n).round(2),
    "Time_Human_AI (min)": np.random.uniform(4, 10, n).round(2)
})

# Display sample data
print("Sample Data:\n")
print(data.head(10))

# Save dataset
data.to_csv("radiology_diagnostic_500_samples.csv", index=False)

# Calculate averages
print("\nAverage Values:\n")
print(data.mean())

# Performance metrics
accuracy_improvement = ((data["Accuracy_Human_AI (%)"].mean() - data["Accuracy_Conventional (%)"].mean())
                        / data["Accuracy_Conventional (%)"].mean()) * 100

error_reduction = ((data["Error_Conventional (%)"].mean() - data["Error_Human_AI (%)"].mean())
                   / data["Error_Conventional (%)"].mean()) * 100

time_reduction = ((data["Time_Conventional (min)"].mean() - data["Time_Human_AI (min)"].mean())
                  / data["Time_Conventional (min)"].mean()) * 100

print("\nPerformance Improvements:\n")
print(f"Accuracy Improvement: {accuracy_improvement:.2f}%")
print(f"Error Reduction: {error_reduction:.2f}%")
print(f"Diagnosis Time Reduction: {time_reduction:.2f}%")

# Plot comparison
labels = ["Accuracy", "Error", "Time"]
conventional = [
    data["Accuracy_Conventional (%)"].mean(),
    data["Error_Conventional (%)"].mean(),
    data["Time_Conventional (min)"].mean()
]

human_ai = [
    data["Accuracy_Human_AI (%)"].mean(),
    data["Error_Human_AI (%)"].mean(),
    data["Time_Human_AI (min)"].mean()
]

x = np.arange(len(labels))

plt.figure()
plt.bar(x - 0.2, conventional, width=0.4, label="Conventional")
plt.bar(x + 0.2, human_ai, width=0.4, label="Human-AI")
plt.xticks(x, labels)
plt.title("Diagnostic Performance Comparison")
plt.legend()
plt.show()
```
