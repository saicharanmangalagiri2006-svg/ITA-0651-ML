```python
import kagglehub
path = kagglehub.dataset_download("ziya07/manufacturing-production-data")
```

```python
import os

# List the contents of the downloaded dataset directory
file_list = os.listdir(path)
print(f"Files in the dataset directory: {file_list}")

# Assuming the main data file is a CSV, let's find it and load it.
# This might need adjustment based on the actual file names.
data_file = None
for f in file_list:
    if f.endswith('.csv'):
        data_file = os.path.join(path, f)
        break

if data_file:
    print(f"Loading data from: {data_file}")
    import pandas as pd
    df = pd.read_csv(data_file)
    print("Displaying the first 5 rows of the dataset:")
    display(df.head())
else:
    print("No CSV file found in the dataset directory. Please specify which file to load.")
```

### Step 1: Initial Data Inspection and Missing Values Check

First, let's display the first few rows of the dataset and check for any missing values or anomalies using `info()` and `isnull().sum()`.

```python
import pandas as pd

print("Dataset Information:")
df.info()

print("\nMissing values per column:")
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0])

print("\nFirst 5 rows of the dataset:")
display(df.head())
```

### Step 2: Handle Missing Values

We will fill numerical columns with their mean and categorical columns with their mode. We'll iterate through all columns to apply this logic.

```python
for column in df.columns:
    if df[column].isnull().any():
        if df[column].dtype in ['int64', 'float64']:
            # Fill numerical missing values with the mean
            df[column].fillna(df[column].mean(), inplace=True)
            print(f"Filled missing numerical values in '{column}' with mean: {df[column].mean():.2f}")
        else:
            # Fill categorical missing values with the mode
            mode_val = df[column].mode()[0]
            df[column].fillna(mode_val, inplace=True)
            print(f"Filled missing categorical values in '{column}' with mode: {mode_val}")

print("\nMissing values after imputation:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

### Step 3: Encode Categorical String Columns

We'll identify categorical columns (object dtype) and apply `LabelEncoder` to convert them into numerical values. For columns with many unique values, one-hot encoding might be more appropriate, but for simplicity and common use cases, LabelEncoder is a good starting point.

```python
from sklearn.preprocessing import LabelEncoder

# Identify categorical columns
categorical_cols = df.select_dtypes(include=['object']).columns

print(f"Categorical columns identified for encoding: {list(categorical_cols)}")

for col in categorical_cols:
    if df[col].nunique() > 0: # Only encode if there are unique values
        le = LabelEncoder()
        df[col] = le.fit_transform(df[col])
        print(f"Encoded column '{col}': {le.classes_} mapped to {le.transform(le.classes_)}")
    else:
        print(f"Skipping encoding for '{col}' as it has no unique values.")

print("\nFirst 5 rows after categorical encoding:")
display(df.head())
```

### Step 4: Scale or Normalize Numerical Efficiency Features

We will scale numerical features using `StandardScaler` to ensure they contribute equally to model training. We will exclude the 'Job_ID' column from scaling as it's likely an identifier.

```python
from sklearn.preprocessing import StandardScaler

# Identify numerical columns for scaling, excluding 'Job_ID' if it exists and is an identifier.
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
if 'Job_ID' in numerical_cols:
    numerical_cols.remove('Job_ID')

print(f"Numerical columns identified for scaling: {numerical_cols}")

if numerical_cols:
    scaler = StandardScaler()
    df[numerical_cols] = scaler.fit_transform(df[numerical_cols])
    print("Numerical features scaled successfully.")
else:
    print("No numerical columns found for scaling.")

print("\nFirst 5 rows after numerical scaling:")
display(df.head())
```

### Step 5: Save the Processed DataFrame

Now, we'll save the cleaned and preprocessed DataFrame to a new CSV file named `processed_robotics_efficiency.csv`.

```python
output_filename = 'processed_robotics_efficiency.csv'
df.to_csv(output_filename, index=False)
print(f"Processed DataFrame saved to '{output_filename}'")
```

# Traffic Network Dataset Analysis

### STEP 1 – LOAD AND SAMPLE

We will load the `processed_traffic_data__1_.csv` file into a pandas DataFrame and then create a random sample of 15 rows from it. Finally, we'll display the sampled DataFrame.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the CSV using pandas
df = pd.read_csv('processed_traffic_data__1_.csv')

# Randomly sample exactly 15 rows
df_sample = df.sample(n=15, random_state=42)

print("First 5 rows of the full dataset:")
display(df.head())

print("\nRandom 15-row sample (df_sample):")
display(df_sample)
```

### STEP 2 – COMPUTE METRICS

For the 5 key columns in the `df_sample`, we will compute the Mean, Standard Deviation, Min, Max, and an 'Accuracy Proxy' (percentage of values > 0). The results will be presented in a clean metrics summary table using a pandas DataFrame.

```python
key_columns = [
    'Traffic Volume (Mbps)', 'Latency (ms)', 'Congestion Level',
    'AI Prediction (Traffic Load)', 'Transaction Success (1/0)'
]

metrics_data = {}

for col in key_columns:
    mean_val = df_sample[col].mean()
    std_val = df_sample[col].std()
    min_val = df_sample[col].min()
    max_val = df_sample[col].max()
    accuracy_proxy = (df_sample[col] > 0).sum() / len(df_sample) * 100

    metrics_data[col] = {
        'Mean': mean_val,
        'Standard Deviation': std_val,
        'Min': min_val,
        'Max': max_val,
        'Accuracy Proxy (>0%)': accuracy_proxy
    }

metrics_df = pd.DataFrame.from_dict(metrics_data, orient='index')

print("Metrics Summary Table for the 15-row Sample:")
display(metrics_df.round(2))
```

### STEP 3 – MATPLOTLIB VISUALIZATIONS

Now, we will create a figure with 2 rows and 3 columns of subplots to visualize different aspects of the data using `matplotlib` only, as specified. This includes bar charts, line charts, scatter plots, a histogram, and a pie chart.

```python
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle("Traffic Network Dataset – Statistical Analysis", fontsize=16)

# Define the 5 key columns again for clarity
key_columns = [
    'Traffic Volume (Mbps)', 'Latency (ms)', 'Congestion Level',
    'AI Prediction (Traffic Load)', 'Transaction Success (1/0)'
]

# Plot 1 – Bar Chart: Mean values of the 5 key columns from the 15-row sample
mean_values_sample = df_sample[key_columns].mean()
axes[0, 0].bar(mean_values_sample.index, mean_values_sample.values, color='skyblue')
axes[0, 0].set_title('Mean Values (15-row Sample)')
axes[0, 0].tick_params(axis='x', rotation=45)
for i, v in enumerate(mean_values_sample.values):
    axes[0, 0].text(i, v + 0.05, f'{v:.2f}', ha='center', va='bottom', fontsize=8)
axes[0, 0].grid(axis='y', linestyle='--', alpha=0.7)

# Plot 2 – Line Chart: 'Traffic Volume (Mbps)' vs 'AI Prediction (Traffic Load)' for the 15 sample rows
axes[0, 1].plot(df_sample.index, df_sample['Traffic Volume (Mbps)'], marker='o', label='Traffic Volume (Mbps)')
axes[0, 1].plot(df_sample.index, df_sample['AI Prediction (Traffic Load)'], marker='x', label='AI Prediction (Traffic Load)')
axes[0, 1].set_title('Traffic Volume vs AI Prediction (15-row Sample)')
axes[0, 1].set_xlabel('Sample Index')
axes[0, 1].set_ylabel('Value')
axes[0, 1].legend()
axes[0, 1].grid(True, linestyle='--', alpha=0.7)

# Plot 3 – Scatter Plot: 'Latency (ms)' on X-axis vs 'Congestion Level' on Y-axis. Color points by 'Transaction Success (1/0)'
scatter = axes[0, 2].scatter(
    df_sample['Latency (ms)'], df_sample['Congestion Level'],
    c=df_sample['Transaction Success (1/0)'], cmap='viridis', alpha=0.7
)
axes[0, 2].set_title('Latency vs Congestion Level (15-row Sample)')
axes[0, 2].set_xlabel('Latency (ms)')
axes[0, 2].set_ylabel('Congestion Level')
cbar = fig.colorbar(scatter, ax=axes[0, 2])
cbar.set_label('Transaction Success (1/0)')
axes[0, 2].grid(True, linestyle='--', alpha=0.7)

# Plot 4 – Histogram: Distribution of 'Latency (ms)' from the FULL dataset with 30 bins
axes[1, 0].hist(df['Latency (ms)'], bins=30, color='lightgreen', edgecolor='black', alpha=0.7)
mean_latency = df['Latency (ms)'].mean()
axes[1, 0].axvline(mean_latency, color='red', linestyle='dashed', linewidth=1, label=f'Mean: {mean_latency:.2f}')
axes[1, 0].set_title('Distribution of Latency (ms) (Full Dataset)')
axes[1, 0].set_xlabel('Latency (ms)')
axes[1, 0].set_ylabel('Frequency')
axes[1, 0].legend()
axes[1, 0].grid(True, linestyle='--', alpha=0.7)

# Plot 5 – Standard Deviation Bar Chart: Show std deviation of all 14 columns from the full dataset
std_full_data = df.std(numeric_only=True)
axes[1, 1].bar(std_full_data.index, std_full_data.values, color='lightcoral')
axes[1, 1].set_title('Standard Deviation of All Columns (Full Dataset)')
axes[1, 1].tick_params(axis='x', rotation=45, ha='right')
axes[1, 1].set_ylabel('Standard Deviation')
axes[1, 1].grid(axis='y', linestyle='--', alpha=0.7)

# Plot 6 – Pie Chart: Count of positive vs negative values in 'Transaction Success (1/0)' from the full dataset
success_counts = df['Transaction Success (1/0)'].value_counts()
labels = ['Success (1)', 'Failure (0)']
colors = ['#66b3ff', '#ff9999']

axes[1, 2].pie(
    success_counts.values, labels=labels, autopct='%1.1f%%', colors=colors,
    startangle=90, wedgeprops={'edgecolor': 'black'}
)
axes[1, 2].set_title('Transaction Success (1/0) (Full Dataset)')
axes[1, 2].axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.

plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Adjust layout to prevent overlap
plt.savefig('traffic_analysis.png')
plt.show()
```

### STEP 4 – ACCURACY METRIC

Finally, we will compute the percentage of rows where 'Transaction Success (1/0)' is greater than 0 for both the 15-row sample and the full dataset, then print these values.

```python
# Compute accuracy on the 15-row sample
sample_accuracy = (df_sample['Transaction Success (1/0)'] > 0).sum() / len(df_sample) * 100

# Compute accuracy on the full dataset
full_data_accuracy = (df['Transaction Success (1/0)'] > 0).sum() / len(df) * 100

print(f"Sample Accuracy  : {sample_accuracy:.2f}%")
print(f"Full Data Accuracy: {full_data_accuracy:.2f}%")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# --- STEP 1: LOAD & SAMPLE ---
print("### STEP 1: Load and Sample ###")

# Load the CSV using pandas
df = pd.read_csv('processed_robotics_efficiency.csv')

# Randomly sample 12 rows (random_state=42)
df_sample = df.sample(n=12, random_state=42)

# Print the sampled dataframe
print("Randomly sampled 12 rows:")
display(df_sample)

# --- STEP 2: COMPUTE METRICS ---
print("\n### STEP 2: Compute Metrics on Sampled Data ###")

numeric_cols = ['Processing_Time', 'Energy_Consumption', 'Machine_Availability', 'Material_Used']

metrics_data = {}
for col in numeric_cols:
    metrics_data[col] = {
        'Mean': df_sample[col].mean(),
        'Standard Deviation': df_sample[col].std(),
        'Min': df_sample[col].min(),
        'Max': df_sample[col].max()
    }

# Accuracy proxy: percentage of sampled rows where Job_Status == 'Completed'
sample_accuracy_proxy = (df_sample['Job_Status'] == 'Completed').sum() / len(df_sample) * 100

metrics_df = pd.DataFrame.from_dict(metrics_data, orient='index').round(2)
metrics_df.loc['Job_Status_Completed_Accuracy'] = [
    sample_accuracy_proxy if c == 'Mean' else np.nan for c in metrics_df.columns
]
metrics_df.rename(index={'Job_Status_Completed_Accuracy': 'Accuracy Proxy (% Completed)'}, inplace=True)

print("Metrics Summary Table for the 12-row Sample:")
display(metrics_df)

# --- STEP 3: MATPLOTLIB GRAPHS ---
print("\n### STEP 3: Matplotlib Graphs ###")

# Use specified style and color palette
plt.style.use('seaborn-v0_8-whitegrid')
colors = ['#4C72B0', '#DD8452', '#55A868', '#C44E52', '#8C96C6', '#FC8D62'] # Extended for more categories

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Robotics Efficiency Dataset Analysis', fontsize=16)

# Plot 1: Bar chart - Mean values of the 4 numeric columns with error bars (std deviation)
mean_numeric = df_sample[numeric_cols].mean()
std_numeric = df_sample[numeric_cols].std()
axes[0, 0].bar(mean_numeric.index, mean_numeric.values, yerr=std_numeric.values, capsize=5, color=colors[:len(numeric_cols)])
axes[0, 0].set_title('Mean Numeric Metrics with Std Dev (Sample)')
axes[0, 0].set_ylabel('Value')
axes[0, 0].tick_params(axis='x', rotation=45, ha='right')

# Plot 2: Box plot - Distribution of all 4 numeric columns side by side
df_sample[numeric_cols].boxplot(ax=axes[0, 1], grid=False, rot=45)
axes[0, 1].set_title('Distribution of Numeric Metrics (Sample)')
axes[0, 1].set_ylabel('Value')

# Plot 3: Pie chart - Count of each Job_Status category in the full dataset
job_status_counts = df['Job_Status'].value_counts()
axes[1, 0].pie(job_status_counts, labels=job_status_counts.index, autopct='%1.1f%%', startangle=90, colors=colors)
axes[1, 0].set_title('Job Status Distribution (Full Dataset)')
axes[1, 0].axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.

# Plot 4: Grouped bar chart - Count of Optimization_Category broken down by Job_Status (Full Dataset)
# Ensure 'Operation_Type' and 'Job_Status' are correctly identified for the full dataset
# Assuming 'Optimization_Category' is the column name for the breakdown

# Ensure 'Job_Status' and 'Optimization_Category' are present in the DataFrame.
if 'Job_Status' in df.columns and 'Optimization_Category' in df.columns:
    grouped_counts = df.groupby(['Job_Status', 'Optimization_Category']).size().unstack(fill_value=0)
    grouped_counts.plot(kind='bar', ax=axes[1, 1], color=colors)
    axes[1, 1].set_title('Optimization Category by Job Status (Full Dataset)')
    axes[1, 1].set_ylabel('Count')
    axes[1, 1].set_xlabel('Job Status')
    axes[1, 1].tick_params(axis='x', rotation=45, ha='right')
    axes[1, 1].legend(title='Optimization Category', bbox_to_anchor=(1.05, 1), loc='upper left')
else:
    axes[1, 1].text(0.5, 0.5, 'Missing Job_Status or Optimization_Category column for grouped bar chart.',
                    horizontalalignment='center', verticalalignment='center', transform=axes[1, 1].transAxes)
    axes[1, 1].set_title('Grouped Bar Chart (Skipped)')

plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Adjust layout to prevent overlap with suptitle
plt.savefig('robotics_analysis.png', dpi=150)
plt.show()

# --- STEP 4: SUMMARY ---
print("\n### STEP 4: Summary ###")

total_dataset_size = len(df)
sample_size = len(df_sample)

# Mean accuracy (% Completed jobs) in the sample (already calculated)
# Which numeric column has the highest std deviation (from the full dataset)
std_dev_full_data = df[numeric_cols].std()
highest_std_col = std_dev_full_data.idxmax()
highest_std_val = std_dev_full_data.max()

print(f"Total dataset size: {total_dataset_size} rows")
print(f"Sample size: {sample_size} rows")
print(f"Mean accuracy (% Completed jobs) in the sample: {sample_accuracy_proxy:.2f}%")
print(f"Numeric column with the highest standard deviation in the full dataset: '{highest_std_col}' ({highest_std_val:.2f})")
```
